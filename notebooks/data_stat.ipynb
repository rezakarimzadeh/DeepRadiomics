```python
import os
import json
import pandas as pd

def read_json(file_path):
    with open(file_path, 'r') as f:
        data = json.load(f)
    return data
```

```python
data_root = "/home/reza/Documents/Reza_projects/08_drarabi_lymphnodes/new_dataset"
dataset_dicts = read_json(os.path.join(data_root, "dataset_directories.json"))
center_1 = dataset_dicts['Masih-SUV']
center_2 = dataset_dicts['Razavi-SUV']
combined_centers = center_1 + center_2
```

```python
def change_to_df(center_data, center_name):
    agg_list = []
    for p_info in center_data:
        radiomics_dir = os.path.join(data_root, center_name, str(list(p_info.keys())[0]), "radiomics", 'radiomics_each_lesion.json')
        nodes_radiomics = read_json(radiomics_dir)
        info = dict(list(p_info.values())[0])
        info['lesion_count'] = len(nodes_radiomics)
        agg_list.append(info)

    center_df = pd.DataFrame(agg_list)
    center_df = center_df.rename(columns={"Gender(Male=0, Fmale:1)": "Gender(Male:0, Female:1)"})
    center_df = center_df.drop(columns=['Stage'])
    return center_df
center_1_df = change_to_df(center_1, center_name='Masih-SUV')
center_2_df = change_to_df(center_2, center_name='Razavi-SUV')
combined_centers_df = pd.concat([center_1_df, center_2_df], ignore_index=True)
```

# age and gender statistical tests

```python
import numpy as np
from scipy import stats

def demographic_stats(df, center_name):
    print(f"\n===== {center_name} =====")

    hl = df[df["Binary Label"] == "HL"]
    nhl = df[df["Binary Label"] == "NHL"]

    # ---------- AGE ----------
    age_hl = hl["Age"].dropna()
    age_nhl = nhl["Age"].dropna()

    # Welch t-test (robust to unequal variance)
    t_stat, p_age = stats.ttest_ind(age_hl, age_nhl, equal_var=False)

    # Cohen's d
    mean1, mean2 = age_hl.mean(), age_nhl.mean()
    std1, std2 = age_hl.std(ddof=1), age_nhl.std(ddof=1)
    n1, n2 = len(age_hl), len(age_nhl)
    pooled_std = np.sqrt(((n1 - 1)*std1**2 + (n2 - 1)*std2**2) / (n1 + n2 - 2))
    cohens_d = (mean1 - mean2) / pooled_std

    print("\nAge comparison (HL vs NHL):")
    print(f"HL mean±std: {mean1:.2f} ± {std1:.2f}")
    print(f"NHL mean±std: {mean2:.2f} ± {std2:.2f}")
    print(f"p-value: {p_age:.4e}")
    print(f"Cohen's d: {cohens_d:.3f}")

    # ---------- GENDER ----------
    gender_hl = hl["Gender(Male:0, Female:1)"]
    gender_nhl = nhl["Gender(Male:0, Female:1)"]

    # contingency table
    table = np.array([
        [(gender_hl == 0).sum(), (gender_hl == 1).sum()],
        [(gender_nhl == 0).sum(), (gender_nhl == 1).sum()]
    ])

    # chi-square test
    chi2, p_gender, _, _ = stats.chi2_contingency(table)

    # Cramér's V
    n = table.sum()
    cramers_v = np.sqrt(chi2 / (n * (min(table.shape) - 1)))

    print("\nGender comparison (HL vs NHL):")
    print("Contingency table (Male, Female):")
    print(table)
    print(f"p-value: {p_gender:.4e}")
    print(f"Cramér's V: {cramers_v:.3f}")

demographic_stats(center_1_df, "Center 1")
demographic_stats(center_2_df, "Center 2")
```

```python
import numpy as np
from scipy.stats import wasserstein_distance
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold, cross_val_score

def quantify_domain_shift(output):
    X1 = np.asarray(output["center1"]["data"], dtype=float)
    X2 = np.asarray(output["center2"]["data"], dtype=float)

    assert X1.shape[1] == X2.shape[1], "Center 1 and Center 2 must have same feature dimension"

    # ---- SMD ----
    mu1, mu2 = X1.mean(axis=0), X2.mean(axis=0)
    sd1, sd2 = X1.std(axis=0, ddof=1), X2.std(axis=0, ddof=1)
    pooled_sd = np.sqrt((sd1**2 + sd2**2) / 2.0)
    pooled_sd[pooled_sd == 0] = 1e-8
    smd = (mu1 - mu2) / pooled_sd
    abs_smd = np.abs(smd)

    # ---- Wasserstein distance ----
    wd = np.array([wasserstein_distance(X1[:, i], X2[:, i]) for i in range(X1.shape[1])])

    # ---- Domain classifier AUC ----
    X = np.vstack([X1, X2])
    y = np.array([0] * len(X1) + [1] * len(X2))
    clf = LogisticRegression(max_iter=2000, class_weight="balanced")
    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    domain_auc = cross_val_score(clf, X, y, cv=cv, scoring="roc_auc").mean()

    # ---- Summary ----
    results = {
        "n_features": X1.shape[1],
        "median_abs_smd": np.median(abs_smd),
        "iqr_abs_smd": np.percentile(abs_smd, 75) - np.percentile(abs_smd, 25),
        "pct_abs_smd_gt_0.2": 100 * np.mean(abs_smd > 0.2),
        "pct_abs_smd_gt_0.5": 100 * np.mean(abs_smd > 0.5),
        "median_wasserstein": np.median(wd),
        "iqr_wasserstein": np.percentile(wd, 75) - np.percentile(wd, 25),
        "domain_auc": domain_auc,
    }

    print("Domain shift summary")
    print(f"Number of features                : {results['n_features']}")
    print(f"Median |SMD|                     : {results['median_abs_smd']:.4f}")
    print(f"IQR |SMD|                        : {results['iqr_abs_smd']:.4f}")
    print(f"% features with |SMD| > 0.2      : {results['pct_abs_smd_gt_0.2']:.2f}")
    print(f"% features with |SMD| > 0.5      : {results['pct_abs_smd_gt_0.5']:.2f}")
    print(f"Median Wasserstein distance      : {results['median_wasserstein']:.4f}")
    print(f"IQR Wasserstein distance         : {results['iqr_wasserstein']:.4f}")
    print(f"Domain-classifier ROC-AUC        : {results['domain_auc']:.4f}")

    return results, abs_smd, wd

# run
results, abs_smd, wd = quantify_domain_shift(aggregated_data_dict)
```
