```python
import pandas as pd
import os
from glob import glob
from pathlib import Path
```

```python
def load_results_table(results_dir, methods_list, use_tda):
    agg_list = []
    for method in methods_list:
        aggregated_result_dir = Path(results_dir)/f"{method}{'_tda_radiomics' if use_tda else '_radiomics'}" / "aggregated_results.json"
        df = pd.read_json(aggregated_result_dir)
        agg_list.append(df)
    final_df = pd.concat(agg_list, ignore_index=True)
    display(final_df)
```

```python
print("Results with Radiomics Features:")
load_results_table("../Results", ["transformer", "mil", "deep_sets", "ml_models"], use_tda=False)
```

```python
print("Results with Radiomics + TDA Features:")
load_results_table("../Results", ["transformer", "mil", "deep_sets", "ml_models"], use_tda=True)
```
