```python
import json
from typing import Any, Union

def load_json(path: str) -> Union[dict, list]:
    """Load a JSON file and return its Python object (usually dict or list)."""
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)

import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc

def plot_cv_roc_compare_fancy(
    method_to_folds,
    title="5-Fold Cross-Validated ROC",
    n_grid=500,
    show_folds=False,
    show_band=True,
    show_chance=True,
    show_iso_tpr=False,
    legend_loc="lower right",
    savepath=None,
    dpi=300,
):
    """
    method_to_folds: dict
      { "Method": [(y_true_fold0, y_score_fold0), ..., (y_true_fold4, y_score_fold4)] }

    Notes:
      - y_score must be continuous score for positive class (probability/logit/decision score).
      - All methods should use the same CV splits for fair comparison.
    """

    # --- Global, clean style (matplotlib-native) ---
    plt.rcParams.update({
        "font.size": 12,
        "axes.titlesize": 14,
        "axes.labelsize": 12,
        "legend.fontsize": 10,
        "axes.spines.top": False,
        "axes.spines.right": False,
    })

    mean_fpr = np.linspace(0.0, 1.0, n_grid)

    # Figure
    fig, ax = plt.subplots(figsize=(8.6, 7.2))
    ax.set_title(title, pad=12)

    # Subtle grid
    ax.grid(True, which="major", alpha=0.25)
    ax.grid(True, which="minor", alpha=0.10)
    ax.minorticks_on()

    # Optional: iso-TPR guide lines (diagonal-ish guides are not standard; keep it optional)
    if show_iso_tpr:
        for t in [0.2, 0.4, 0.6, 0.8]:
            # Not a canonical ROC guide; this is purely aesthetic.
            ax.axhline(t, lw=0.8, alpha=0.15)

    # Chance line
    if show_chance:
        ax.plot([0, 1], [0, 1], linestyle="--", lw=1.2, alpha=0.7, label="Chance")

    # Plot each method
    for method_name, folds in method_to_folds.items():
        tprs = []
        aucs = []

        for y_true, y_score in folds:
            y_true = np.asarray(y_true).astype(int)
            y_score = np.asarray(y_score)

            fpr, tpr, _ = roc_curve(y_true, y_score)
            fold_auc = auc(fpr, tpr)
            aucs.append(fold_auc)

            interp_tpr = np.interp(mean_fpr, fpr, tpr)
            interp_tpr[0] = 0.0
            tprs.append(interp_tpr)

            if show_folds:
                ax.plot(fpr, tpr, lw=1.0, alpha=0.12)

        tprs = np.vstack(tprs)
        mean_tpr = tprs.mean(axis=0)
        std_tpr = tprs.std(axis=0)
        mean_tpr[-1] = 1.0

        mean_auc = float(np.mean(aucs))
        std_auc = float(np.std(aucs))

        # Main mean curve (matplotlib default colors auto-cycle)
        (line,) = ax.plot(
            mean_fpr, mean_tpr, lw=2.8,
            label=f"{method_name}  |  AUC={mean_auc:.3f}±{std_auc:.3f}"
        )

        # Shaded band uses same color as the line
        if show_band:
            color = line.get_color()
            tpr_lower = np.clip(mean_tpr - std_tpr, 0.0, 1.0)
            tpr_upper = np.clip(mean_tpr + std_tpr, 0.0, 1.0)
            ax.fill_between(mean_fpr, tpr_lower, tpr_upper, alpha=0.14, color=color)

    # Axes formatting
    ax.set_xlim(0.0, 1.0)
    ax.set_ylim(0.0, 1.05)
    ax.set_xlabel("False Positive Rate (1 - Specificity)")
    ax.set_ylabel("True Positive Rate (Sensitivity)")

    # Legend with a bit of polish
    leg = ax.legend(loc=legend_loc, frameon=True, framealpha=0.9, borderpad=0.6)
    leg.get_frame().set_linewidth(0.6)

    fig.tight_layout()

    if savepath is not None:
        fig.savefig(savepath, dpi=dpi, bbox_inches="tight")

    plt.show()



def selected_plot_roc_curves(results_dir, methods_list, use_coords, use_demographic, ml_models=["AdaBoost", 'TabPFN'], file_name="results_fold"):
    method_folds = {}
    for method in methods_list:
        if use_demographic and use_coords:
            folder_name = f"{method}_coords_demographic_radiomics"
        elif use_demographic and not use_coords:
            folder_name = f"{method}_demographic_radiomics"
        elif not use_demographic and use_coords:
            folder_name = f"{method}_coords_radiomics"
        else:
            folder_name = f"{method}_radiomics"
        folds = []
        ml_folds = {}
        for i in range(5):
            fold_result_dir = Path(results_dir)/folder_name /f"fold_{i}" /f'{file_name}_{i}.json'
            fold_results = load_json(fold_result_dir)
            # choose only ml models for ROC curves
            if method == "ml_models":
                for mm in ml_models:
                    if mm not in ml_folds:
                        ml_folds[mm] = []
                    ml_folds[mm].append([fold_results[mm]['y_true'], fold_results[mm]['y_prob']])
            else:
                folds.append([fold_results['y_true'], fold_results['y_prob']])
        if method != "ml_models":   
            method_folds[method] = folds
    method_folds.update(ml_folds)
    # Now plot ROC curves
    plot_cv_roc_compare_fancy(method_folds, title="5-Fold CV ROC Comparison")


selected_plot_roc_curves("../Results", models, use_coords=True, use_demographic=True)
    
```

```python
def plot_cv_roc_compare_fancy(
    method_to_folds,
    title="5-Fold Cross-Validated ROC",
    n_grid=500,
    show_folds=False,
    show_band=True,
    show_chance=True,
    show_iso_tpr=False,
    legend_loc="lower right",
    savepath=None,
    dpi=300,
    ax=None,               # NEW
):
    plt.rcParams.update({
        "font.size": 12,
        "axes.titlesize": 14,
        "axes.labelsize": 12,
        "legend.fontsize": 10,
        "axes.spines.top": False,
        "axes.spines.right": False,
    })

    mean_fpr = np.linspace(0.0, 1.0, n_grid)

    created_fig = False
    if ax is None:
        fig, ax = plt.subplots(figsize=(8.6, 7.2))
        created_fig = True
    else:
        fig = ax.figure

    ax.set_title(title, pad=12)
    ax.grid(True, which="major", alpha=0.25)
    ax.grid(True, which="minor", alpha=0.10)
    ax.minorticks_on()

    if show_iso_tpr:
        for t in [0.2, 0.4, 0.6, 0.8]:
            ax.axhline(t, lw=0.8, alpha=0.15)

    if show_chance:
        ax.plot([0, 1], [0, 1], linestyle="--", lw=1.2, alpha=0.7, label="Chance")

    for method_name, folds in method_to_folds.items():
        tprs = []
        aucs = []

        for y_true, y_score in folds:
            y_true = np.asarray(y_true).astype(int)
            y_score = np.asarray(y_score)

            fpr, tpr, _ = roc_curve(y_true, y_score)
            fold_auc = auc(fpr, tpr)
            aucs.append(fold_auc)

            interp_tpr = np.interp(mean_fpr, fpr, tpr)
            interp_tpr[0] = 0.0
            tprs.append(interp_tpr)

            if show_folds:
                ax.plot(fpr, tpr, lw=1.0, alpha=0.12)

        tprs = np.vstack(tprs)
        mean_tpr = tprs.mean(axis=0)
        std_tpr = tprs.std(axis=0)
        mean_tpr[-1] = 1.0

        mean_auc = float(np.mean(aucs))
        std_auc = float(np.std(aucs))

        (line,) = ax.plot(
            mean_fpr, mean_tpr, lw=2.8,
            label=f"{method_name}  |  AUC={mean_auc:.3f}±{std_auc:.3f}"
        )

        if show_band:
            color = line.get_color()
            tpr_lower = np.clip(mean_tpr - std_tpr, 0.0, 1.0)
            tpr_upper = np.clip(mean_tpr + std_tpr, 0.0, 1.0)
            ax.fill_between(mean_fpr, tpr_lower, tpr_upper, alpha=0.14, color=color)

    ax.set_xlim(0.0, 1.0)
    ax.set_ylim(0.0, 1.05)
    ax.set_xlabel("False Positive Rate (1 - Specificity)")
    ax.set_ylabel("True Positive Rate (Sensitivity)")

    leg = ax.legend(loc=legend_loc, frameon=True, framealpha=0.9, borderpad=0.6)
    leg.get_frame().set_linewidth(0.6)

    if savepath is not None:
        fig.savefig(savepath, dpi=dpi, bbox_inches="tight")

    if created_fig:
        fig.tight_layout()
        plt.show()
from pathlib import Path  # make sure this exists

def comparision_plot_roc_curves(
    results_dir,
    methods_list,
    use_coords,
    use_demographic,
    ml_models=["AdaBoost", "TabPFN"],
    file_name="results_fold",
):
    # helper that runs your existing loading logic for a specific (coords, demographic) setting
    def build_method_folds(use_coords_local, use_demographic_local):
        method_folds = {}
        ml_folds = {}

        for method in methods_list:
            if use_demographic_local and use_coords_local:
                folder_name = f"{method}_coords_demographic_radiomics"
            elif use_demographic_local and not use_coords_local:
                folder_name = f"{method}_demographic_radiomics"
            elif not use_demographic_local and use_coords_local:
                folder_name = f"{method}_coords_radiomics"
            else:
                folder_name = f"{method}_radiomics"

            folds = []
            for i in range(5):
                fold_result_dir = Path(results_dir) / folder_name / f"fold_{i}" / f"{file_name}_{i}.json"
                fold_results = load_json(fold_result_dir)

                if method == "ml_models":
                    for mm in ml_models:
                        ml_folds.setdefault(mm, [])
                        ml_folds[mm].append([fold_results[mm]["y_true"], fold_results[mm]["y_prob"]])
                else:
                    folds.append([fold_results["y_true"], fold_results["y_prob"]])

            if method != "ml_models":
                method_folds[method] = folds

        method_folds.update(ml_folds)
        return method_folds

    # ---- 2x2 subplots over the 4 folder variants ----
    fig, axs = plt.subplots(2, 2, figsize=(16, 12), sharex=True, sharey=True)

    configs = [
        (True,  True,  "coords + demographic + radiomics"),
        (False, True,  "demographic + radiomics"),
        (True,  False, "coords + radiomics"),
        (False, False, "radiomics only"),
    ]

    for ax, (c, d, subtitle) in zip(axs.ravel(), configs):
        method_folds = build_method_folds(use_coords_local=c, use_demographic_local=d)
        plot_cv_roc_compare_fancy(
            method_folds,
            title=subtitle,
            ax=ax,
            show_folds=False,
            show_band=True,
            show_chance=True,
            legend_loc="lower right",
        )

    # fig.suptitle("5-Fold CV ROC Comparison (All Feature Configurations)", y=1.02, fontsize=16)
    fig.tight_layout()
    fig.savefig("figures/roc_comparison_center1.png", dpi=300, bbox_inches="tight")

    plt.show()

comparision_plot_roc_curves("../Results", models, use_coords=True, use_demographic=True)
```

```python
from pathlib import Path

def build_metric_by_config_from_json(
    results_dir,
    methods_list,
    metric_key="f1_score",
    ml_models=("AdaBoost", "TabPFN"),
    file_name="results_fold",
):
    """
    Collects a scalar metric per fold from your JSON files.

    Returns:
      config_to_method_vals: dict
        {
          "coords+demo+radiomics": {"deep_sets":[..5..], "TabPFN":[..5..], ...},
          "demo+radiomics": {...},
          "coords+radiomics": {...},
          "radiomics": {...},
        }
    """

    configs = [
        (True,  True,  "coords+demo+radiomics"),
        (False, True,  "demo+radiomics"),
        (True,  False, "coords+radiomics"),
        (False, False, "radiomics"),
    ]

    config_to_method_vals = {}

    for use_coords, use_demographic, cfg_name in configs:
        method_vals = {}
        ml_vals = {}

        for method in methods_list:
            # same folder naming as your code
            if use_demographic and use_coords:
                folder_name = f"{method}_coords_demographic_radiomics"
            elif use_demographic and not use_coords:
                folder_name = f"{method}_demographic_radiomics"
            elif not use_demographic and use_coords:
                folder_name = f"{method}_coords_radiomics"
            else:
                folder_name = f"{method}_radiomics"

            for i in range(5):
                fold_path = Path(results_dir) / folder_name / f"fold_{i}" / f"{file_name}_{i}.json"
                fold_results = load_json(str(fold_path))

                if method == "ml_models":
                    for mm in ml_models:
                        if metric_key not in fold_results.get(mm, {}):
                            raise KeyError(
                                f"Missing metric '{metric_key}' in fold_results['{mm}'] at: {fold_path}"
                            )
                        ml_vals.setdefault(mm, [])
                        ml_vals[mm].append(float(fold_results[mm][metric_key]))
                else:
                    if metric_key not in fold_results:
                        raise KeyError(
                            f"Missing metric '{metric_key}' in fold_results at: {fold_path}"
                        )
                    method_vals.setdefault(method, [])
                    method_vals[method].append(float(fold_results[metric_key]))

        method_vals.update(ml_vals)
        config_to_method_vals[cfg_name] = method_vals
        # reorder for consistent plotting
        config_to_method_vals[cfg_name] = {k: method_vals[k] for k in ['transformer', 'deep_sets', 'mil', 'set_transformer', 'graph', 'AdaBoost', 'TabPFN']}
    return config_to_method_vals

def plot_grouped_metric_boxplots(
    config_to_method_vals,
    metric_name="F1-score",
    title=None,
    group_order=("coords+demo+radiomics", "demo+radiomics", "coords+radiomics", "radiomics"),
    ylim=None,
    figsize=(16, 7),
    savepath=None,
    dpi=300,
):
    # all_methods = sorted({m for cfg in config_to_method_vals.values() for m in cfg.keys()})
    all_methods = list(config_to_method_vals[group_order[0]].keys())
    plt.rcParams.update({
        "font.size": 12,
        "axes.titlesize": 14,
        "axes.labelsize": 12,
        "legend.fontsize": 10,
        "axes.spines.top": False,
        "axes.spines.right": False,
    })

    # if title is None:
        # title = f"{metric_name} (5-Fold CV) by Feature Configuration"

    fig, ax = plt.subplots(figsize=figsize)
    ax.set_title(title, pad=12)
    ax.set_ylabel(metric_name)

    if ylim is not None:
        ax.set_ylim(*ylim)

    ax.grid(True, axis="y", alpha=0.25)
    ax.minorticks_on()
    ax.grid(True, which="minor", axis="y", alpha=0.10)

    box_w = 0.55
    inner_gap = 0.22
    group_gap = 1.2

    cmap = plt.get_cmap("tab10")
    method_to_color = {m: cmap(i % 10) for i, m in enumerate(all_methods)}

    positions, data, box_colors = [], [], []
    positions_methods = []
    group_centers = []
    group_boundaries = []
    x = 0.0

    for cfg_name in group_order:
        start = x

        for method in all_methods:
            vals = config_to_method_vals.get(cfg_name, {}).get(method, [])
            if len(vals) == 0:
                x += (box_w + inner_gap)
                continue

            positions.append(x)
            data.append(vals)
            box_colors.append(method_to_color[method])
            positions_methods.append(method)
            x += (box_w + inner_gap)

        end = x - (box_w + inner_gap)
        group_centers.append((start + end) / 2.0)
        group_boundaries.append(x + group_gap / 2.0)
        x += group_gap

    bp = ax.boxplot(
        data,
        positions=positions,
        widths=box_w,
        patch_artist=True,
        showfliers=False,
        medianprops=dict(linewidth=2.0),
        whiskerprops=dict(linewidth=1.5),
        capprops=dict(linewidth=1.5),
        boxprops=dict(linewidth=1.5),
    )

    for patch, c in zip(bp["boxes"], box_colors):
        patch.set_facecolor(c)
        patch.set_alpha(0.22)

    # jittered fold points
    rng = np.random.default_rng(0)
    for pos, vals, c in zip(positions, data, box_colors):
        vals = np.asarray(vals, dtype=float)
        jitter = rng.normal(loc=0.0, scale=0.055, size=len(vals))
        ax.scatter(
            np.full_like(vals, pos) + jitter,
            vals,
            s=28,
            alpha=0.8,
            edgecolors="none",
            color=c,
            zorder=3,
        )

    # Bottom axis: 4 configuration labels
    ax.set_xticks(group_centers)
    ax.set_xticklabels(group_order)
    ax.set_xlabel("Feature configuration")

    for boundary in group_boundaries[:-1]:
        ax.axvline(boundary, lw=1.0, alpha=0.12)

    # Top axis: method labels aligned to each box
    ax_top = ax.twiny()
    ax_top.set_xlim(ax.get_xlim())
    ax_top.set_xticks(positions)
    ax_top.set_xticklabels(positions_methods, rotation=60, ha="left")
    ax_top.set_xlabel("Models")
    ax_top.tick_params(axis="x", length=0, pad=6)

    # Legend (optional; remove if you prefer)
    handles = [
        plt.Line2D([0], [0], marker="s", linestyle="", markersize=10,
                   markerfacecolor=method_to_color[m], alpha=0.6, label=m)
        for m in all_methods
    ]
    # ax.legend(handles=handles, title="Methods", loc="upper right", frameon=True, framealpha=0.9)

    fig.tight_layout()

    if savepath is not None:
        fig.savefig(savepath, dpi=dpi, bbox_inches="tight")

    plt.show()

vals = build_metric_by_config_from_json(
    results_dir="../Results",
    methods_list=models,
    metric_key="f1_score",
    ml_models=("AdaBoost", "TabPFN"),
)
plot_grouped_metric_boxplots(vals, metric_name="F1-score", ylim=(0.6, 1.02), savepath="figures/f1_score_boxplot_center1.png")

vals = build_metric_by_config_from_json(
    results_dir="../Results",
    methods_list=models,
    metric_key="roc_auc",
    ml_models=("AdaBoost", "TabPFN"),
)
plot_grouped_metric_boxplots(vals, metric_name="ROC-AUC", ylim=(0.6, 1.02), savepath="figures/roc_auc_boxplot_center1.png")
vals = build_metric_by_config_from_json(
    results_dir="../Results",
    methods_list=models,
    metric_key="accuracy",
    ml_models=("AdaBoost", "TabPFN"),
)
plot_grouped_metric_boxplots(vals, metric_name="Accuracy", ylim=(0.6, 1.02), savepath="figures/accuracy_boxplot_center1.png")
```

```python
from pathlib import Path
import json
import numpy as np
import matplotlib.pyplot as plt


def load_json(path: str):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def plot_methodwise_grouped_versions_boxplots_from_folds(
    results_dir,
    deep_methods,
    metric_key="f1_score",
    folds=(0, 1, 2, 3, 4),
    group_order=("coords+demo+radiomics", "demo+radiomics", "coords+radiomics", "radiomics"),
    # group -> suffix exactly matching your folders
    group_to_suffix=None,
    # version name -> filename template inside fold folder
    versions=(
        ("Base (center2)", "center2_results_fold_{i}.json"),
        ("TTA-Norm",       "tta_norm_center2_results_fold_{i}.json"),
        ("TENT",           "tta_tent_center2_results_fold_{i}.json"),
    ),
    file_name="results_fold",  # not used here (kept for compatibility if you extend)
    file_must_exist=True,
    figsize_per_method=5.5,
    ylim=(0, 1.02),
    show_points=True,
    show_legend=True,
    add_separators=True,
    separator_style=None,
    savepath=None,
    dpi=300,
):
    """
    Expected structure like:
      Results/
        deep_sets_demographic_radiomics/
          fold_0/
            center2_results_fold_0.json
            tta_norm_center2_results_fold_0.json
            tta_tent_center2_results_fold_0.json
            ...
          fold_1/...
        deep_sets_coords_demographic_radiomics/...

    Plot:
      1 x N_methods subplots
      Each subplot: 4 groups on x-axis; within each group 3 boxplots (Base/TTA-Norm/TENT)
      Vertical "fancy" separator lines between group blocks (after each 3-box block).
    """

    if group_to_suffix is None:
        group_to_suffix = {
            "coords+demo+radiomics": "coords_demographic_radiomics",
            "demo+radiomics":        "demographic_radiomics",
            "coords+radiomics":      "coords_radiomics",
            "radiomics":             "radiomics",
        }

    # Default separator style
    if separator_style is None:
        separator_style = dict(
            linewidth=2.0,
            linestyles=(0, (4, 6)),  # dashed: 4 on, 6 off
            alpha=0.18,
            zorder=0
        )

    n_methods = len(deep_methods)
    fig, axes = plt.subplots(
        1, n_methods,
        figsize=(figsize_per_method * n_methods, 6.6),
        sharey=True
    )
    if n_methods == 1:
        axes = [axes]

    plt.rcParams.update({
        "font.size": 12,
        "axes.titlesize": 14,
        "axes.labelsize": 12,
        "legend.fontsize": 10,
        "axes.spines.top": False,
        "axes.spines.right": False,
    })

    # Layout: per group, 3 boxes
    n_versions = len(versions)
    base_positions = np.arange(len(group_order))  # group centers: 0..3
    group_width = 0.80
    box_w = group_width / (n_versions + 0.8)
    offsets = np.linspace(-group_width/2 + box_w/2, group_width/2 - box_w/2, n_versions)

    # Colors per version
    cmap = plt.get_cmap("tab10")
    version_colors = {vname: cmap(i % 10) for i, (vname, _) in enumerate(versions)}

    rng = np.random.default_rng(0)

    def folder_for(method, group):
        return f"{method}_{group_to_suffix[group]}"

    def file_for(version_template, i):
        return version_template.format(i=i)

    # One shared legend for the whole figure
    legend_handles = [
        plt.Line2D([0], [0], marker="s", linestyle="", markersize=10,
                   markerfacecolor=version_colors[vname], alpha=0.6, label=vname)
        for vname, _ in versions
    ]

    for ax, method in zip(axes, deep_methods):
        ax.set_title(method)
        ax.grid(True, axis="y", alpha=0.25)
        ax.minorticks_on()
        ax.grid(True, which="minor", axis="y", alpha=0.10)

        all_data = []
        all_pos = []
        all_cols = []
        group_boundaries = []  # x positions for separators between group blocks

        # Collect data in order: group-major, version-minor
        for gi, group in enumerate(group_order):
            folder = Path(results_dir) / folder_for(method, group)

            group_pos = []  # x positions of the 3 version boxes in this group

            for vi, (vname, vtemplate) in enumerate(versions):
                vals = []
                for i in folds:
                    fold_dir = folder / f"fold_{i}"
                    fpath = fold_dir / file_for(vtemplate, i)

                    if not fpath.exists():
                        if file_must_exist:
                            raise FileNotFoundError(
                                f"Missing file: {fpath}\n"
                                f"Expected under: {fold_dir}\n"
                                f"Check templates in `versions=`."
                            )
                        else:
                            continue

                    d = load_json(str(fpath))
                    if metric_key not in d:
                        raise KeyError(f"Missing '{metric_key}' in {fpath}")

                    vals.append(float(d[metric_key]))

                pos = base_positions[gi] + offsets[vi]
                group_pos.append(pos)

                all_data.append(vals)
                all_pos.append(pos)
                all_cols.append(version_colors[vname])

            # boundary between this group and next group (except last group)
            if gi < len(group_order) - 1:
                rightmost = max(group_pos)
                leftmost_next = base_positions[gi + 1] + min(offsets)
                boundary_x = (rightmost + leftmost_next) / 2.0
                group_boundaries.append(boundary_x)

        bp = ax.boxplot(
            all_data,
            positions=all_pos,
            widths=box_w,
            patch_artist=True,
            showfliers=False,
            medianprops=dict(linewidth=2.0),
            whiskerprops=dict(linewidth=1.4),
            capprops=dict(linewidth=1.4),
            boxprops=dict(linewidth=1.4),
        )

        for patch, c in zip(bp["boxes"], all_cols):
            patch.set_facecolor(c)
            patch.set_alpha(0.22)

        if show_points:
            for pos, vals, c in zip(all_pos, all_data, all_cols):
                vals = np.asarray(vals, dtype=float)
                jitter = rng.normal(loc=0.0, scale=box_w * 0.12, size=len(vals))
                ax.scatter(
                    np.full_like(vals, pos) + jitter,
                    vals,
                    s=26,
                    alpha=0.82,
                    edgecolors="none",
                    color=c,
                    zorder=3,
                )

        # Fancy vertical separators between 3-box blocks (group boundaries)
        if add_separators and len(group_boundaries) > 0:
            # Ensure y-lims are set before drawing full-height separators
            if ylim is not None:
                ax.set_ylim(*ylim)
            y0, y1 = ax.get_ylim()
            for xsep in group_boundaries:
                ax.vlines(xsep, y0, y1, **separator_style)

        ax.set_xticks(base_positions)
        ax.set_xticklabels(group_order, rotation=18, ha="right")
        ax.set_xlabel("Feature configuration")

        if ylim is not None:
            ax.set_ylim(*ylim)

    axes[0].set_ylabel(metric_key)

    if show_legend:
        fig.legend(
            handles=legend_handles,
            title="Version",
            loc="lower center",
            ncol=len(versions),
            frameon=True,
            framealpha=0.9,
            bbox_to_anchor=(0.5, 1.03),
        )

    fig.tight_layout()

    if savepath is not None:
        fig.savefig(savepath, dpi=dpi, bbox_inches="tight")

    plt.show()




plot_methodwise_grouped_versions_boxplots_from_folds(
    results_dir="../Results",
    deep_methods=new_models,
    metric_key="f1_score",             
    savepath="figures/center2_tta_f1score.png",
)

plot_methodwise_grouped_versions_boxplots_from_folds(
    results_dir="../Results",
    deep_methods=new_models,
    metric_key="roc_auc",             
    savepath="figures/center2_tta_rocauc.png",
)
```

```python
from pathlib import Path
import json
import numpy as np
import matplotlib.pyplot as plt

from scipy.stats import ttest_rel


def load_json(path: str):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


def p_to_stars(p: float) -> str:
    if not np.isfinite(p):
        return ""
    if p < 1e-4:
        return "****"
    if p < 1e-3:
        return "***"
    if p < 1e-2:
        return "**"
    if p < 5e-2:
        return "*"
    return ""  # IMPORTANT: no 'ns'


def holm_adjust(pvals):
    pvals = np.asarray(pvals, dtype=float)
    finite = np.isfinite(pvals)
    out = np.full_like(pvals, np.nan, dtype=float)
    if finite.sum() == 0:
        return out

    idx = np.where(finite)[0]
    p = pvals[finite]
    order = np.argsort(p)
    p_sorted = p[order]

    adj_sorted = np.empty_like(p_sorted)
    running_max = 0.0
    m = len(p_sorted)
    for i, p_i in enumerate(p_sorted):
        factor = (m - i)
        running_max = max(running_max, factor * p_i)
        adj_sorted[i] = running_max

    adj_sorted = np.minimum(adj_sorted, 1.0)

    inv = np.empty_like(order)
    inv[order] = np.arange(m)
    out[idx] = adj_sorted[inv]
    return out


def paired_t_pvalue(a, b, alternative="two-sided"):
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)

    if len(a) == 0 or len(b) == 0:
        return np.nan
    if len(a) != len(b):
        raise ValueError(f"Paired t-test requires equal length, got {len(a)} vs {len(b)}")

    res = ttest_rel(a, b)
    p_two = float(res.pvalue)

    if alternative == "two-sided":
        return p_two

    diff_mean = float(np.mean(a - b))
    p_one = p_two / 2.0

    if alternative == "greater":
        return p_one if diff_mean > 0 else 1.0 - p_one
    if alternative == "less":
        return p_one if diff_mean < 0 else 1.0 - p_one

    raise ValueError("alternative must be 'two-sided', 'greater', or 'less'")


def add_sig_bracket(ax, x1, x2, y, h, text, lw=1.2, alpha=0.9, fontsize=14):
    ax.plot([x1, x1, x2, x2], [y, y + h, y + h, y],
            lw=lw, alpha=alpha, color="black", clip_on=False)
    ax.text((x1 + x2) / 2, y + h, text,
            ha="center", va="bottom", fontsize=fontsize, color="black", clip_on=False)


def plot_methodwise_grouped_versions_boxplots_from_folds(
    results_dir,
    deep_methods,
    metric_key="f1_score",
    folds=(0, 1, 2, 3, 4),
    group_order=("coords+demo+radiomics", "demo+radiomics", "coords+radiomics", "radiomics"),
    group_to_suffix=None,
    versions=(
        ("Base (center2)", "center2_results_fold_{i}.json"),
        ("TTA-Norm",       "tta_norm_center2_results_fold_{i}.json"),
        ("TENT",           "tta_tent_center2_results_fold_{i}.json"),
    ),
    file_must_exist=True,
    figsize_per_method=5.5,
    ylim=(0, 1.02),
    show_points=True,
    show_legend=True,
    add_separators=True,
    separator_style=None,
    # --- p-value annotations (paired t-test + Holm) ---
    add_pvalues=False,
    alpha=0.05,
    pvalue_pairs=((0, 1), (0, 2), (1, 2)),
    pvalue_alternative="two-sided",     # "two-sided", "greater", or "less"
    holm_scope="within_group",          # "within_group" or "within_method"
    pvalue_bracket_height=0.010,        # fraction of y-range
    pvalue_step=0.045,                  # fraction of y-range
    pvalue_fontsize=14,
    savepath=None,
    dpi=300,
):
    if group_to_suffix is None:
        group_to_suffix = {
            "coords+demo+radiomics": "coords_demographic_radiomics",
            "demo+radiomics":        "demographic_radiomics",
            "coords+radiomics":      "coords_radiomics",
            "radiomics":             "radiomics",
        }

    if separator_style is None:
        separator_style = dict(
            linewidth=2.0,
            linestyles=(0, (4, 6)),
            alpha=0.18,
            zorder=0
        )

    n_methods = len(deep_methods)
    fig, axes = plt.subplots(
        1, n_methods,
        figsize=(figsize_per_method * n_methods, 6.6),
        sharey=True
    )
    if n_methods == 1:
        axes = [axes]

    plt.rcParams.update({
        "font.size": 12,
        "axes.titlesize": 14,
        "axes.labelsize": 12,
        "legend.fontsize": 10,
        "axes.spines.top": False,
        "axes.spines.right": False,
    })

    n_versions = len(versions)
    base_positions = np.arange(len(group_order))
    group_width = 0.80
    box_w = group_width / (n_versions + 0.8)
    offsets = np.linspace(-group_width/2 + box_w/2, group_width/2 - box_w/2, n_versions)

    cmap = plt.get_cmap("tab10")
    version_colors = {vname: cmap(i % 10) for i, (vname, _) in enumerate(versions)}
    rng = np.random.default_rng(0)

    def folder_for(method, group):
        return f"{method}_{group_to_suffix[group]}"

    def file_for(version_template, i):
        return version_template.format(i=i)

    legend_handles = [
        plt.Line2D([0], [0], marker="s", linestyle="", markersize=10,
                   markerfacecolor=version_colors[vname], alpha=0.6, label=vname)
        for vname, _ in versions
    ]

    for ax, method in zip(axes, deep_methods):
        ax.set_title(method)
        ax.grid(True, axis="y", alpha=0.25)
        ax.minorticks_on()
        ax.grid(True, which="minor", axis="y", alpha=0.10)

        all_data, all_pos, all_cols = [], [], []
        group_boundaries = []

        group_version_vals = {g: {} for g in group_order}
        group_version_pos  = {g: {} for g in group_order}

        for gi, group in enumerate(group_order):
            folder = Path(results_dir) / folder_for(method, group)
            group_pos = []

            for vi, (vname, vtemplate) in enumerate(versions):
                vals = []
                for i in folds:
                    fold_dir = folder / f"fold_{i}"
                    fpath = fold_dir / file_for(vtemplate, i)

                    if not fpath.exists():
                        if file_must_exist:
                            raise FileNotFoundError(f"Missing file: {fpath}")
                        else:
                            continue

                    d = load_json(str(fpath))
                    if metric_key not in d:
                        raise KeyError(f"Missing '{metric_key}' in {fpath}")
                    vals.append(float(d[metric_key]))

                pos = base_positions[gi] + offsets[vi]
                group_pos.append(pos)

                all_data.append(vals)
                all_pos.append(pos)
                all_cols.append(version_colors[vname])

                group_version_vals[group][vi] = vals
                group_version_pos[group][vi] = pos

            if gi < len(group_order) - 1:
                rightmost = max(group_pos)
                leftmost_next = base_positions[gi + 1] + min(offsets)
                group_boundaries.append((rightmost + leftmost_next) / 2.0)

        bp = ax.boxplot(
            all_data,
            positions=all_pos,
            widths=box_w,
            patch_artist=True,
            showfliers=False,
            medianprops=dict(linewidth=2.0),
            whiskerprops=dict(linewidth=1.4),
            capprops=dict(linewidth=1.4),
            boxprops=dict(linewidth=1.4),
        )

        for patch, c in zip(bp["boxes"], all_cols):
            patch.set_facecolor(c)
            patch.set_alpha(0.22)

        if show_points:
            for pos, vals, c in zip(all_pos, all_data, all_cols):
                vals = np.asarray(vals, dtype=float)
                jitter = rng.normal(loc=0.0, scale=box_w * 0.12, size=len(vals))
                ax.scatter(
                    np.full_like(vals, pos) + jitter,
                    vals,
                    s=26,
                    alpha=0.82,
                    edgecolors="none",
                    color=c,
                    zorder=3,
                )

        if ylim is not None:
            ax.set_ylim(*ylim)

        if add_separators and len(group_boundaries) > 0:
            y0, y1 = ax.get_ylim()
            for xsep in group_boundaries:
                ax.vlines(xsep, y0, y1, **separator_style)

        ax.set_xticks(base_positions)
        ax.set_xticklabels(group_order, rotation=18, ha="right")
        ax.set_xlabel("Feature configuration")

        # ---- ONLY annotate significant comparisons (no ns, no brackets for ns) ----
        if add_pvalues:
            y0, y1 = ax.get_ylim()
            y_range = y1 - y0

            # raw p-values per group
            raw_p_by_group = {}
            for group in group_order:
                raw = []
                for (a_idx, b_idx) in pvalue_pairs:
                    a = group_version_vals[group].get(a_idx, [])
                    b = group_version_vals[group].get(b_idx, [])
                    p = paired_t_pvalue(a, b, alternative=pvalue_alternative)
                    raw.append(p)
                raw_p_by_group[group] = raw

            # Holm adjustment
            if holm_scope == "within_group":
                adj_p_by_group = {g: holm_adjust(raw_p_by_group[g]) for g in group_order}
            elif holm_scope == "within_method":
                flat, back = [], []
                for g in group_order:
                    for k, p in enumerate(raw_p_by_group[g]):
                        flat.append(p)
                        back.append((g, k))
                flat_adj = holm_adjust(flat)
                adj_p_by_group = {g: [np.nan] * len(pvalue_pairs) for g in group_order}
                for adj, (g, k) in zip(flat_adj, back):
                    adj_p_by_group[g][k] = adj
            else:
                raise ValueError("holm_scope must be 'within_group' or 'within_method'")

            # draw only significant
            for group in group_order:
                max_in_group = max(np.max(group_version_vals[group][vi]) for vi in group_version_vals[group])
                base_y = max_in_group + 0.04 * y_range
                level = 0  # stack only significant ones

                for (a_idx, b_idx), p_adj in zip(pvalue_pairs, adj_p_by_group[group]):
                    if not np.isfinite(p_adj) or p_adj >= alpha:
                        continue  # SKIP non-significant

                    stars = p_to_stars(p_adj)
                    if stars == "":
                        continue

                    x1 = group_version_pos[group][a_idx]
                    x2 = group_version_pos[group][b_idx]
                    y = base_y + level * (pvalue_step * y_range)

                    add_sig_bracket(
                        ax, x1, x2,
                        y=y,
                        h=pvalue_bracket_height * y_range,
                        text=stars,
                        lw=1.1,
                        alpha=0.9,
                        fontsize=pvalue_fontsize,
                    )
                    level += 1  # only increment when we draw something

    axes[0].set_ylabel(metric_key)

    if show_legend:
        fig.legend(
            handles=legend_handles,
            title="Version",
            loc="upper center",
            ncol=len(versions),
            frameon=True,
            framealpha=0.9,
            bbox_to_anchor=(0.5, 1.03),
        )

    fig.tight_layout()

    if savepath is not None:
        fig.savefig(savepath, dpi=dpi, bbox_inches="tight")

    plt.show()



plot_methodwise_grouped_versions_boxplots_from_folds(
    results_dir="../Results",
    deep_methods=new_models,
    metric_key="f1_score",
    add_pvalues=True,
    pvalue_pairs=((0, 1), (0, 2), (1,2)),     # compare vs Base only (cleaner)
    pvalue_alternative="less",      # if you call paired_t_pvalue(improved, base); see below note
    holm_scope="within_group",
    savepath="figures/f1_boxplots_ttest_holm_center2.png",
)

plot_methodwise_grouped_versions_boxplots_from_folds(
    results_dir="../Results",
    deep_methods=new_models,
    metric_key="roc_auc",
    add_pvalues=True,
    pvalue_pairs=((0, 1), (0, 2), (1,2)),     # compare vs Base only (cleaner)
    pvalue_alternative="less",      # if you call paired_t_pvalue(improved, base); see below note
    holm_scope="within_group",
    savepath="figures/roc_auc_boxplots_ttest_holm_center2.png",
)
```
